```python
df["payment"] = df["payment"].fillna(df["payment"].mode()[0])
```

```python
df["pickup_zone"] = df["pickup_zone"].fillna(df["pickup_zone"].mode()[0])
df["dropoff_zone"] = df["dropoff_zone"].fillna(df["dropoff_zone"].mode()[0])
```

```python
df["pickup_borough"] = df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
df["dropoff_borough"] = df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])
```

```python
df.isnull().sum()
```

```python
import pandas as pd
df["pickup"] = pd.to_datetime(df["pickup"])
df["pickup_date"] = df["pickup"].dt.date
df["pickup_time"] = df["pickup"].dt.time

df[["pickup", "pickup_date", "pickup_time"]].head()
df["dropoff"] = pd.to_datetime(df["dropoff"])
df["dropoff_date"] = df["dropoff"].dt.date
df["dropoff_time"] = df["dropoff"].dt.time
```

```python
df["pickup_date"] = pd.to_datetime(df["pickup_date"], errors="coerce")
df["dropoff_date"] = pd.to_datetime(df["dropoff_date"], errors="coerce")

df["pickup_time"] = pd.to_datetime(df["pickup_time"], errors="coerce")
df["dropoff_time"] = pd.to_datetime(df["dropoff_time"], errors="coerce")

print(df["pickup_date"].dtype)
print(df["pickup_time"].dtype)
print(df["dropoff_date"].dtype)
print(df["dropoff_time"].dtype)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df["pickup"] = pd.to_datetime(df["pickup"], errors="coerce")

df = df.sort_values("pickup")

plt.figure(figsize=(15, 6))
plt.plot(df["pickup"], df["fare"], color ="green")

plt.xlabel("Pickup Timestamp")
plt.ylabel("Fare")
plt.title("Fare Over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Bar Chart
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Group by pickup_borough and sum the fare
borough_fare = df.groupby("pickup_borough")["fare"].sum().sort_values(ascending=False)

# Plot
plt.figure(figsize=(9, 5))
borough_fare.plot(kind="bar", color="green")
plt.title("Total Fare by Pickup Borough", fontsize=13, fontweight="bold")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare ($)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#Piechart
payment_counts = df["payment"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(payment_counts.values, labels=payment_counts.index, shadow =True,autopct="%1.1f%%",
        startangle=90)
plt.title("Distribution of Trips by Payment Method")
plt.tight_layout()
plt.show()
```

```python

#Histogram
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset("taxis")
plt.figure(figsize=(10, 7))
plt.hist(df["distance"], bins=40, color="green", edgecolor="black", alpha=0.85)
plt.title("Distribution of Trip Distance")
plt.xlabel("Distance (miles)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
#Boxplot
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.boxplot(data=df, x="pickup_borough", y="tip")
plt.title("Tip Amount Distribution by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip ($)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#Count Plot
plt.figure(figsize=(10, 7))
sns.countplot(data=df, x="pickup_borough", order=df["pickup_borough"].value_counts().index)
plt.title("Number of Trips per Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Trip Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#Scatter plot
plt.figure(figsize=(10, 7))
sns.scatterplot(data=df, x="distance", y="fare", hue= "pickup_borough", alpha=0.6, s=25)
plt.title("Number of Trips per Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Trip Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#heatmap
corr_cols = ["distance", "fare", "tip", "tolls", "total"]
corr_matrix = df[corr_cols].corr()

plt.figure(figsize=(7, 6))
sns.heatmap(corr_matrix, annot=True, fmt=".3f", cmap="coolwarm", square=True, linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# pairplot
top_zones = df["pickup_zone"].value_counts().nlargest(5).index
subset = df[df["pickup_zone"].isin(top_zones)]

g = sns.pairplot(subset, vars=["distance", "fare", "tip", "total"], hue="pickup_zone",
                  diag_kind="kde", plot_kws={"alpha": 0.6, "s": 25})
g.fig.suptitle("Pairwise Relationships by Pickup Zone", y=1.02)
plt.show()
```

```python
#Violinplot
plt.figure(figsize=(10, 7))

sns.violinplot(
    data=df,
    x='payment',
    y='fare',
    palette='Pastel1'
)

plt.title('Fare Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare ($)')

plt.show()
```
